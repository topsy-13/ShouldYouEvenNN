```python
import pandas as pd
import json
import os
```

```python
path_forecast = '../experiments/forecast_only'
forecast_files = os.listdir(path_forecast)
forecast_paths = [col for col in forecast_files if col.endswith('.json')]
```

```python
# Make it a df
forecast_results = []
for fcst_result in forecast_paths:
    with open(os.path.join(path_forecast, fcst_result), 'r') as f:
        tmp_fcst = json.load(f) 
        print(tmp_fcst)
        forecast_results.append(tmp_fcst)
```

```python
exp_df = pd.DataFrame(forecast_results)
exp_df.head()
```

```python
forecast_performance = exp_df.describe().loc[['mean', 'std']].T.drop(['dataset_id', 'n_models'])
forecast_performance.to_latex(
    '../analysis_outputs/fcst_performance.tex',
    index=True,
    float_format="%.4f",     # 4 decimal places
    caption="Summary statistics of the EBE-NAS results.",
    label="tab:ebe_summary",
    column_format="lrrrrr",  # adjust to your columns
    escape=False             # allow LaTeX symbols in data
)

```

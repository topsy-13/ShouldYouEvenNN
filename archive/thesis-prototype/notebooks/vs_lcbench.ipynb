```python
import pandas as pd
```

```python
lcbench_path = '../experiments/lcbench/all_datasets_benchmark.csv'
lcbench_df = pd.read_csv(lcbench_path)
```

```python
lcbench_df.head()
```

```python
import ast
```

```python
lcbench_df['lc_time'] = lcbench_df['time'].apply(lambda x: ast.literal_eval(x)[-1])
lcbench_df.head()
```

```python
accuracies_key = ['final_val_accuracy', 'lc_time']

grouping_dict = {
    'final_val_accuracy': 'max',
    'lc_time': 'sum'
}

grouped_lc = lcbench_df[['dataset_name'] + accuracies_key].groupby('dataset_name').agg(grouping_dict) / 100
grouped_lc = grouped_lc.reset_index()
grouped_lc.head()
```

```python
import json
json_path = '../experiments/datasets/openml_datasets2.json'
with open(json_path, 'r') as f:
    dict_map = json.load(f)
dict_map 
```

```python
grouped_lc['dataset_id'] = grouped_lc['dataset_name'].map(dict_map)
grouped_lc.head()
```

```python
syenn_path = '../experiments/testing/final_results/summary/all_experiments.csv'
syenn_df = pd.read_csv(syenn_path)
syenn_df.head()
```

```python
tmp_Df = syenn_df[['ebe_final_surpassed_baseline_internal', 'ebe_expected_utility']].copy()
tmp_Df['u_above_0'] = tmp_Df['ebe_expected_utility'] > 0
tmp_Df[['ebe_final_surpassed_baseline_internal', 'u_above_0']].value_counts().reset_index()
```

```python
joined_df = pd.merge(syenn_df[['data_id', 'baseline_val_acc', 'ebe_final_val_acc_internal', 'ebe_surpassed_val_baseline', 'ebe_time_used_s']], 
                     grouped_lc, how='right', left_on='data_id', right_on='dataset_id'
                     )
joined_df['ebe_surpassesLCBench'] = joined_df['ebe_final_val_acc_internal'] > joined_df['final_val_accuracy']
joined_df['LCBenchSurpassed'] = joined_df['final_val_accuracy'] > joined_df['baseline_val_acc']
joined_df['differenceLCBench'] = joined_df['ebe_final_val_acc_internal'] - joined_df['final_val_accuracy']
joined_df.head()
```

```python
joined_df.to_csv('../experiments/testing/final_results/summary/syenn_vs_lcbench.csv', index=False)
```

```python
joined_df.describe()
```

```python
# Make boxplot comparing accuracies
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization of validation accuracies for ShouldYouEvenNN and LCBench
plt.figure(figsize=(6, 5))

# Reshape data for comparative plotting
melted_df = joined_df.melt(
    value_vars=['ebe_final_val_acc_internal', 'final_val_accuracy'],
    var_name='Method',
    value_name='Validation Accuracy'
)

# Construct comparative boxplot
sns.boxplot(
    x='Method',
    y='Validation Accuracy',
    data=melted_df,
    palette='Set2',
    width=0.6
)

# Academic-style labeling and title
plt.title('Validation Acc. between ShouldYouEvenNN and LCBench', fontsize=13)
plt.xlabel('')
plt.ylabel('Validation Accuracy', fontsize=12)
plt.xticks(
    ticks=[0, 1],
    labels=['ShouldYouEvenNN', 'LCBench (Best Acc.)'],
    fontsize=10
)

plt.grid(axis='y', linestyle='--', linewidth=0.5, alpha=0.7)
plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(7, 6))
sns.scatterplot(
    data=joined_df,
    x="lc_time",
    y="ebe_time_used_s",
    s=80,
    color="#2a7b9b",
    edgecolor="black"
)

# Diagonal line of equal runtime
max_time = max(joined_df["lc_time"].max(), joined_df["ebe_time_used_s"].max())
plt.plot([1e-2, max_time], [1e-2, max_time], "--", color="gray", linewidth=1.2)

plt.xscale("log")
plt.yscale("log")
plt.xlabel("LCBench Runtime (s)", fontsize=12)
plt.ylabel("ShouldYouEvenNN Runtime (s)", fontsize=12)
plt.title("Runtime Comparison: ShouldYouEvenNN vs LCBench", fontsize=13, pad=15)

# Annotate efficiency ratio
median_ratio = (joined_df["lc_time"] / joined_df["ebe_time_used_s"]).median()
plt.text(0.05, 0.9, f"Median Speed-up ≈ ×{median_ratio:.1f}", 
         transform=plt.gca().transAxes, fontsize=11, color="black")

plt.tight_layout()
plt.show()
```
